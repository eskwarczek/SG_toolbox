
# Wprowadzenie do tworzenia własnych toolboxów w ArcGIS Pro

## Wstęp

Tworzenie własnych toolboxów pozwala na automatyzację rutynowych zadań GIS oraz udostępnienie skryptów w przyjaznej, graficznej formie użytkownikom, którzy nie muszą znać Pythona. Dzięki temu można zoptymalizować procesy analizy, mapowania czy zarządzania danymi, korzystając zarówno z wbudowanych narzędzi ArcGIS Pro, jak i własnych rozwiązań.

> *Informacje w tym wprowadzeniu opierają się m.in. na treściach książki **Python for ArcGIS Pro Automate cartography and data analysis using ArcPy, ArcGIS API for Python, Notebooks, and pandas** autorstwa Silas'a Toms oraz Bill'a Parker oraz dokumentacji ArcGIS Pro.*

## Co to jest Toolbox?

**Toolbox** w ArcGIS Pro to zbiór narzędzi (toolów), w którym każdy element może być:
- Narzędziem geoprzetwarzania (geoprocessing tool)
- Narzędziem skryptowym (script tool) – opartym na skryptach w Pythonie

Dzięki toolboxom można:
- Ujednolicić i zautomatyzować procesy analityczne
- Udostępnić gotowe narzędzia innym użytkownikom
- Przeprowadzać iteracyjne operacje na zbiorach danych



## Opisy sekcji narzędzia skryptowego

### 1. Script tool General settings

**Ogólne ustawienia narzędzia skryptowego** definiują podstawowe właściwości, które określają, jak narzędzie będzie prezentowane w toolboxie. W tej sekcji ustawia się między innymi:

- **Nazwa i alias** – wyświetlana nazwa narzędzia oraz krótki alias, który może ułatwić identyfikację.
- **Kategoria** – grupowanie narzędzia w odpowiedniej kategorii, co wpływa na jego organizację w toolboxie.
- **Opis** – krótki opis funkcjonalności narzędzia, umożliwiający użytkownikowi szybkie zrozumienie, do czego służy dane narzędzie.
- **Lokalizacja skryptu** – wskazanie ścieżki do pliku .py, który zawiera logikę narzędzia.

Te ustawienia decydują o tym, jak narzędzie zostanie wyświetlone w interfejsie ArcGIS Pro, a także pomagają w jego późniejszej identyfikacji i użyciu przez innych użytkowników.

---

### 2. Script tool Parameters tab

W zakładce **Parameters tab** definiuje się wszystkie parametry, które narzędzie będzie przyjmować podczas wykonywania. Kluczowe elementy to:

- **Typ danych** – określenie rodzaju danych dla każdego parametru (np. liczba, tekst, warstwa, tabela itp.).
- **Kolejność parametrów** – ustalenie, w jakiej kolejności będą wyświetlane parametry, co ułatwia wprowadzanie danych.
- **Etykiety i opisy** – przyjazne nazwy oraz dodatkowe opisy, które informują użytkownika, jakie dane należy wprowadzić.
- **Domyślne wartości** – ustawienie wartości początkowych, które mogą być zmieniane przez użytkownika.
- **Filtry** – ograniczenie wyboru dostępnych wartości (np. określony typ pliku lub warstwy), co zwiększa precyzję wprowadzanych danych.

Prawidłowo skonfigurowane parametry są kluczem do poprawnego działania narzędzia, zapewniając zarówno elastyczność, jak i łatwość obsługi.

---

### 3. Script tool Validation

Sekcja **Script tool Validation** umożliwia dynamiczne sprawdzanie poprawności wprowadzanych przez użytkownika danych przed uruchomieniem głównej logiki narzędzia. Kluczowe aspekty to:

- **Klasa walidacyjna** – implementacja własnej klasy (często dziedziczącej po klasie narzędzia) umożliwiającej sprawdzanie, modyfikowanie oraz aktualizację listy dostępnych parametrów.
- **Logika warunkowa** – w zależności od wybranych wartości parametrów, można zmieniać dostępność lub kolejność innych parametrów.
- **Komunikaty o błędach i ostrzeżeniach** – generowanie komunikatów, które informują użytkownika o nieprawidłowościach (np. błędnych typach danych, braku wymaganych wartości).

Dzięki walidacji narzędzie może reagować na potencjalne błędy już na etapie konfiguracji, co znacząco ułatwia późniejsze wykonywanie analizy.

---

### 4. Writing messages

Sekcja **Writing messages** opisuje metody komunikacji narzędzia ze swoim użytkownikiem podczas wykonywania. Ważne elementy to:

- **Informacyjne komunikaty** – funkcja `arcpy.AddMessage()` pozwala na wyświetlanie zwykłych komunikatów informacyjnych.
- **Ostrzeżenia** – funkcja `arcpy.AddWarningMessage()` umożliwia przekazywanie ostrzeżeń, które nie przerywają wykonania narzędzia, ale sygnalizują potencjalne problemy.
- **Błędy** – funkcja `arcpy.AddErrorMessage()` jest używana do zgłaszania krytycznych błędów, które mogą przerwać działanie narzędzia.

Dzięki tym metodom użytkownik jest na bieżąco informowany o postępie przetwarzania, a także o ewentualnych problemach, co ułatwia diagnostykę i ewentualne modyfikacje skryptu.


## Tworzenie narzędzia skryptowego

Poniżej przedstawiono kroki niezbędne do stworzenia własnego narzędzia skryptowego w ArcGIS Pro:

### 1. Przygotowanie skryptu w Python

Napisz skrypt, który wykonuje określoną operację. Przykładowo, poniższy skrypt pobiera dwie liczby jako parametry, oblicza ich sumę i wyświetla wynik:

```python
import arcpy

def main():
    # Pobranie parametrów narzędzia
    liczba1 = float(arcpy.GetParameterAsText(0))
    liczba2 = float(arcpy.GetParameterAsText(1))
    
    # Obliczenie sumy
    suma = liczba1 + liczba2
    
    # Przekazanie wyniku jako komunikat
    arcpy.AddMessage(f"Wynik dodawania: {suma}")
    
if __name__ == '__main__':
    main()
```

### 2. Utworzenie nowego Toolboxa w ArcGIS Pro

- W katalogu projektu kliknij prawym przyciskiem myszy → **New** → **Toolbox**.
- Nadaj toolboxowi odpowiednią nazwę (np. `MojeNarzędzia.tbx`).

### 3. Dodanie narzędzia skryptowego do toolboxa

- Kliknij prawym przyciskiem myszy na utworzonym toolboxie i wybierz **New** → **Script**.
- Podaj nazwę narzędzia (np. `DodajLiczby`) oraz opis.
- W sekcji **Source** wskaż wcześniej przygotowany skrypt (np. `dodaj_liczby.py`) lub edytuj skrypt bezpośrednio.

### 4. Konfiguracja parametrów narzędzia

W kreatorze ustaw:
- **Input Parameters** – dodaj dwa parametry typu *Double* (lub *Float*) dla liczb wejściowych.

> Upewnij się, że typy danych parametrów odpowiadają oczekiwanym wartościom w skrypcie.

### 5. Walidacja i testowanie narzędzia

- Uruchom narzędzie z poziomu ArcGIS Pro, wypełniając wartości parametrów.
- Sprawdź komunikaty w oknie narzędzia.
- Dokonaj poprawek w skrypcie lub ustawieniach parametrów, jeśli zajdzie taka potrzeba.



## Przykładowy scenariusz zastosowania

Załóżmy, że chcemy stworzyć narzędzie, które:
- Pobiera warstwę wejściową z danymi
- Przeprowadza operację buforowania na obiektach (np. parków) o zadanej odległości
- Wybiera obiekty (np. przystanki autobusowe) znajdujące się w buforze
- Eksportuje wynik jako nową warstwę

W tym przypadku:
1. Przygotowujesz skrypt, który wykorzystuje narzędzia ArcPy, takie jak `arcpy.analysis.Buffer` i `arcpy.management.SelectLayerByLocation`.
2. Dodajesz skrypt jako narzędzie do toolboxa.
3. Definiujesz parametry wejściowe: ścieżka do warstwy, wartość buforowania, itp.
4. Testujesz narzędzie i weryfikujesz wynik.

## Python Toolbox vs. Standardowy Toolbox

### 1. Struktura i edycja
- **Standardowy Toolbox (custom toolbox):**
  - Składa się z pliku z rozszerzeniem **.atbx**, w którym narzędzia (script tools) są dodawane przy pomocy kreatora.
  - Definicje parametrów, walidacja oraz skrypt źródłowy są rozdzielone – skrypt jest przechowywany jako osobny plik **.py**, a parametry definiowane przez interfejs.
  - Edycja parametrów odbywa się przy użyciu kreatora narzędzi w ArcGIS Pro, co ułatwia pracę osobom początkującym.

- **Python Toolbox:**
  - To pojedynczy plik z rozszerzeniem **.pyt**, który jest czystym kodem Pythona.
  - W ramach jednego pliku definiuje się zarówno strukturę toolboxa, jak i wszystkie narzędzia – ich parametry, logikę wykonania oraz walidację.
  - Wszystkie elementy edytuje się w dowolnym edytorze tekstu lub IDE, co umożliwia łatwiejszą kontrolę wersji i lepszą integrację z narzędziami developerskimi.

### 2. Dynamiczność i elastyczność
- **Standardowy Toolbox:**
  - Korzysta z ustalonych ustawień z kreatora, co sprawia, że konfiguracja jest bardziej statyczna.
  - Walidacja parametrów i logika wykonania są często ograniczone do tego, co pozwala na interfejs narzędzia, co może być niewystarczające przy bardziej złożonych operacjach.

- **Python Toolbox:**
  - Pozwala na dynamiczne definiowanie parametrów – ich właściwości, listy wartości czy zależności między parametrami można ustawiać bezpośrednio w kodzie.
  - Umożliwia wdrożenie bardziej zaawansowanej logiki walidacyjnej, np. dynamiczne zmiany dostępności parametrów w zależności od poprzednich wyborów.
  - Użytkownik z doświadczeniem w Pythonie ma większą swobodę i kontrolę nad narzędziem.

### 3. Przenośność i kontrola wersji
- **Standardowy Toolbox:**
  - Plik **.tbx** jest formatem binarnym, co może utrudniać zarządzanie wersjami przy użyciu systemów kontroli wersji (np. Git).
  - Skrypt źródłowy jest osobnym plikiem **.py**, więc można go śledzić, ale cały kontekst narzędzia jest podzielony.

- **Python Toolbox:**
  - Jako czysty plik tekstowy (.pyt) łatwiej jest śledzić zmiany i zarządzać wersjami całego narzędzia jako jednego pliku.
  - Dzięki temu rozwiązanie jest bardziej przenośne między członkami zespołu.

### 4. Wsparcie dla dodatkowych funkcjonalności
- **Standardowy Toolbox:**
  - Obsługuje różne typy narzędzi – oprócz narzędzi skryptowych można w nim umieszczać też modele utworzone w ModelBuilderze czy narzędzia systemowe.
  - Integracja opiera się głównie na interfejsie użytkownika, co może ograniczać elastyczność w niektórych przypadkach.

- **Python Toolbox:**
  - Skoncentrowany wyłącznie na narzędziach opartych na Pythonie, co umożliwia pełną integrację logiki narzędzia w jednym miejscu.
  - Umożliwia m.in. implementację metody **isLicensed**, która pozwala na dynamiczne sprawdzanie dostępności odpowiednich licencji lub rozszerzeń przed uruchomieniem narzędzia.

---

## Podsumowanie

**Standardowe toolboxy** (custom toolboxes) są idealne dla początkujących – kreator narzędzi ułatwia definiowanie parametrów i tworzenie walidacji, ale cała struktura narzędzia jest rozdzielona między interfejs a osobny plik skryptu.  
**Python toolboxy** to rozwiązanie dla zaawansowanych użytkowników, którzy chcą mieć pełną kontrolę nad wszystkimi aspektami narzędzia w jednym pliku tekstowym, co ułatwia zarządzanie kodem, wersjowanie i dynamiczne definiowanie zachowań narzędzia.

Wybór między nimi zależy głównie od poziomu doświadczenia w Pythonie, potrzeb w zakresie dynamicznej walidacji parametrów oraz wymagań dotyczących kontroli wersji i przenośności.

# Opis projektu zaliczeniowego – Backup danych z AGOL

**Cel projektu:**  
Celem projektu jest stworzenie własnego (customowego) toolboxu w ArcGIS Pro zawierającego narzędzie do tworzenia backupu danych z ArcGIS Online (AGOL) aktualnie zalogowanego użytkownika. Backup polega na pobraniu wszystkich (i tylko) hosted feature layers, zapisaniu ich jako geobazy, a następnie spakowaniu tej geobazy w plik ZIP, którego nazwa automatycznie zawiera datę wykonania backupu. Można albo zapisać wszystkie warstwy w jednej geobazie, ewentualnie pakować każdą warstwę w osobną geobazę.

**Forma pracy:**  
Praca w grupach.

## Wymagania funkcjonalne (Implementacja podstawowej funkcjonalności – 6 punktów)

1. **Utworzenie nowego toolboxu:**
   - Utworzenie  toolboxu w ArcGIS Pro oraz dodanie do niego nowego narzędzia (obojętne jaki typ toolboxu).
   - **(2 punkty)** za prawidłowe utworzenie i integrację narzędzia oraz przejrzystą konfigurację (opis parametrów, nazwy które mają sens itp)

2. **Narzędzie backupu danych:**
   - Narzędzie musi przyjmować ścieżkę do folderu, w którym zostanie zapisany backup.
   - Pobieranie ma dotyczyć wyłącznie feature services przypisanych do aktualnie zalogowanego użytkownika (użycie np. funkcji ArcGIS API for Python/GIS("home")).
   - **(2 punkty)** za prawidłowe pobranie ścieżki oraz wykonanie operacji backupu.

3. **Realizacja operacji backupu:**
   - Backup polega na pobraniu danych do geobazy (np. .gdb) oraz spakowaniu jej w plik ZIP z automatycznie generowaną nazwą zawierającą datę wykonania backupu (np. Backup_AGOL_YYYY-MM-DD.zip).
   - **(2 punkty)** za automatyczne generowanie nazwy z datą oraz poprawne wykonanie tworzenia i kompresji geobazy.
